> **Importante:** a lógica aprendida no WebAlgo continua válida. O que muda principalmente é a forma de escrever o algoritmo.


````python
# @title
from unittest.mock import patch
from io import StringIO
import sys
import math

def _run(func, entradas):
    """Executa a função simulando valores informados por input()."""
    with patch("builtins.input", side_effect=[str(x) for x in entradas]):
        out = StringIO()
        old = sys.stdout
        sys.stdout = out
        try:
            func()
        finally:
            sys.stdout = old
    return out.getvalue().strip()

def _igual_item(recebido, esperado, tol=1e-6):
    try:
        return math.isclose(
            float(recebido),
            float(esperado),
            rel_tol=tol,
            abs_tol=tol
        )
    except (ValueError, TypeError):
        return str(recebido).strip() == str(esperado).strip()

def validar(func, testes, tol=1e-6):
    """Valida saídas simples ou múltiplos valores separados por espaços/linhas."""
    ok = 0

    for entradas, esperado in testes:
        try:
            saida = _run(func, entradas)

            if isinstance(esperado, (list, tuple)):
                recebidos = saida.split()
                esperados = [str(item) for item in esperado]

                passou = (
                    len(recebidos) == len(esperados)
                    and all(
                        _igual_item(recebido, item_esperado, tol)
                        for recebido, item_esperado in zip(recebidos, esperados)
                    )
                )
            else:
                passou = _igual_item(saida, esperado, tol)

            if passou:
                ok += 1
        except Exception:
            pass

    if ok == len(testes):
        print("✅ VALIDADO")
    else:
        print(f"❌ NÃO VALIDADO ({ok}/{len(testes)} testes)")
````

# 1 — Relembrando: WebAlgo → Python

No WebAlgo, vocês já trabalhavam com **variáveis, entrada, processamento, saída e decisões**.  
Em Python, os mesmos conceitos aparecem com uma sintaxe diferente.

| Ideia | WebAlgo | Python |
|---|---|---|
| Mostrar informação | `escreva(...)` | `print(...)` |
| Ler informação | `leia(x)` | `x = input(...)` |
| Inteiro | variável inteira | `int(...)` |
| Real | variável real | `float(...)` |
| Resto da divisão | operador de resto | `%` |
| Divisão inteira | operador de divisão inteira | `//` |
| Potência | potência | `**` |
| Igualdade em uma condição | `=` | `==` |
| Diferente | `<>` | `!=` |
| Se | `se ... entao` | `if ...:` |
| Senão | `senao` | `else:` |

Durante a aula, procure sempre identificar:

**ENTRADA → PROCESSAMENTO → SAÍDA**

# 3 — Da sequência para a decisão

Até agora, todas as instruções eram executadas de cima para baixo.

Mas alguns problemas exigem que o computador **escolha** o que fazer.

Antes do `if`, precisamos compreender que uma comparação produz um valor lógico:

- `True` — verdadeiro;
- `False` — falso.


````python
a = 10
b = 7

print(a > b)
print(a < b)
print(a == b)
print(a != b)
print(a >= 10)
print(b <= 5)
````

## Muito importante: `=` não é `==`

Em Python:

```python
idade = 16
```

significa **atribuir** o valor `16` à variável `idade`.

Já:

```python
idade == 16
```

é uma **comparação** e produz `True` ou `False`.

Essa diferença será fundamental nos algoritmos condicionais.


````python
idade = 16

print(idade)
print(idade == 16)
print(idade == 18)
````

## Operadores relacionais

| Operador | Significado |
|---|---|
| `==` | igual |
| `!=` | diferente |
| `>` | maior |
| `<` | menor |
| `>=` | maior ou igual |
| `<=` | menor ou igual |

Agora vamos usar essas comparações dentro de `if`.


# 4 — Primeiro `if`

No WebAlgo, a ideia era:

```text
se condição entao
    comandos
senao
    comandos
fimse
```

Em Python:

```python
if condição:
    comandos
else:
    comandos
```

Observe duas diferenças importantes:

1. há `:` depois da condição;
2. o bloco é identificado pela **indentação**.

Python não usa `fimse`.


````python
idade = int(input("Digite sua idade: "))

if idade >= 18:
    print("Maior de idade")
else:
    print("Menor de idade")
````

## `elif`: quando existem mais de duas possibilidades

```python
if condição1:
    ...
elif condição2:
    ...
else:
    ...
```

O Python testa as condições **na ordem**. Quando encontra uma verdadeira, executa aquele bloco e não testa os seguintes.


````python
numero = float(input("Digite um número: "))

if numero > 0:
    print("Positivo")
elif numero < 0:
    print("Negativo")
else:
    print("Zero")
````
